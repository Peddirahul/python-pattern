```python
for i in range(1,6):
    for j in range(1,6):
        print("*",end="")
    print()
```

```python

for i in range(1,6):
    for j in range(1,6):
        print(i, end="")
    print()
```

```python
for i in range(1,6):
    for j in range(1,6):
        print(j,end="")
    print()
```

```python
for i in range(ord('A'),ord('E')+1):
    for j in range(ord('A'),ord('E')+1):
        print(chr(i),end="")
    print()
```

```python
for i in range(ord('A'),ord('E')+1):
    for j in range(ord('A'),ord('E')+1):
        print(chr(j),end="")
    print()
              
```
