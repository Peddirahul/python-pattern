```python
for i in range(1,6):
    print(" "*(6-i)*2,end= "")
    print((str(i)+ " ")*i)
```

```python
for i in range(1,6):
    print(" "*(6-i)*2,end= "")
    for j in range(1,i+1):
        print(j,end= " ")
    print()
```

```python
for i in range(1,6):
    print(" "*(6-i)*2,end="")
    for j in range(1,i+1):
        print(chr(64+i),end= " ")
    print()
```
